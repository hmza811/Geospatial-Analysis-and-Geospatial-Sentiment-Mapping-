# TASK 2: Geospatial Sentiment Analysis Using Social Media Data

```python
# Import Libraries

import geopandas as gpd
import pandas as pd
import pandas as pd
import matplotlib.pyplot as plt
from textblob import TextBlob
# Import Nominatim
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
```

```python
df_tweets = pd.read_csv('C:/Users/HP/Downloads/tweets.csv')
df_tweets.head()
```

# Task 2.1: Data Pre-processin

```python
# Keep only rows where 'user_location' is not null
df_filtered = df_tweets[df_tweets['user_location'].notna()]
df_filtered
```

```python
import re

# Combine regex patterns for better performance
CLEAN_TWEET_PATTERN = re.compile(
    r'http\S+|@\w+|#|[^\w\s]|[\U00010000-\U0010ffff]', 
    flags=re.UNICODE
)

def clean_tweet(tweet):
    return CLEAN_TWEET_PATTERN.sub('', tweet).lower()
df_cleantweets = df_filtered[df_filtered['user_location'].notna()].copy()

# Now apply the cleaning function without worrying about the warning
df_cleantweets['tweets'] = df_cleantweets['tweets'].apply(clean_tweet)
df_cleantweets
```

```python
df_cleantweets_sample = df_cleantweets.sample(n=600,random_state= 1)
df_cleantweets_sample
```

# Task 2.2: Geocoding

```python
geolocator = Nominatim(user_agent="CIS7030")
```

```python
geocode = RateLimiter(geolocator.geocode, min_delay_seconds=1)
```

```python
df_cleantweets_sample['location'] = df_cleantweets_sample['user_location'].apply(geocode)
df_cleantweets_sample['Lat'] = df_cleantweets_sample['location'].apply(lambda x: x.latitude if x else None)
df_cleantweets_sample['Lon'] = df_cleantweets_sample['location'].apply(lambda x: x.longitude if x else None)
df_cleantweets_sample 
```

```python
 # First, check for rows where 'location' is NaN
nan_locations = df_cleantweets_sample['location'].isna()

# Filter the DataFrame to exclude rows with NaN in 'location'
df_filtered2 = df_cleantweets_sample[~nan_locations]
df_filtered2
```

# TASK 2.3 & 2.4 Polarity & Subjectivity Anallysis

```python
def getTextPolarity(txt):
    return TextBlob(txt).sentiment.polarity
```

```python
def getTextSubjectivity(txt):
    return TextBlob(txt).sentiment.subjectivity
```

```python
df_filtered['polarity'] = df_filtered['tweets'].apply(getTextPolarity)
df_filtered['subjectivity'] = df_filtered['tweets'].apply(getTextSubjectivity)
df_filtered.head()
```

```python
 # Plot histogram for polarity
plt.figure(figsize=(10, 5))
plt.hist(df_filtered['polarity'], bins=20, color='green', alpha=0.7)
plt.title('Polarity Histogram')
plt.xlabel('Polarity')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

The most significant peak is at the polarity value of 0, with a frequency of around 3500, indicating that a large number of tweets are neutral in sentiment. This suggests that many users are discussing ChatGPT in a factual or neutral manner without strong positive or negative emotions.
 There are smaller peaks at positive polarity values, particularly between 0 and 0.5, indicating that a moderate number of tweets express mildly positive sentiments. Fewer tweets have strongly positive sentiments, as shown by the lower frequencies near the polarity value of 1.

Negative sentiments are much less frequent, with relatively few tweets having polarity values below 0. The frequencies gradually decrease as the polarity becomes more negative, suggesting that negative sentiments about ChatGPT are less common.

In summary, the histogram indicates that discussions about ChatGPT on Twitter are predominantly neutral, with a significant portion of tweets expressing mildly positive sentiments and relatively few tweets being negative.

```python
df_filtered2['polarity'] = df_filtered2['tweets'].apply(getTextPolarity)

# Categorize polarity
def categorize_polarity(pol):
    if pol < 0:
        return 'Negative'
    elif pol > 0:
        return 'Positive'
    else:
        return 'Neutral'

df_filtered2['category'] = df_filtered2['polarity'].apply(categorize_polarity)

# Calculate average polarity by location
df_grouped = df_filtered2.groupby(['Lat', 'Lon', 'category']).agg({'polarity': 'mean'}).reset_index()

# Create a GeoDataFrame
geometry = [Point(xy) for xy in zip(df_grouped['Lon'], df_grouped['Lat'])]
gdf = gpd.GeoDataFrame(df_grouped, geometry=geometry)

# Load world map data
world = gpd.read_file(gpd.datasets.get_path('naturalearth_lowres'))

# Plotting
fig, ax = plt.subplots(figsize=(15, 10))
world.plot(ax=ax, color='lightgray')

# Custom color map
category_colors = {'Negative': 'red', 'Neutral': 'yellow', 'Positive': 'green'}
gdf['color'] = gdf['category'].map(category_colors)

# Plot the points
for category, color in category_colors.items():
    subset = gdf[gdf['category'] == category]
    subset.plot(ax=ax, markersize=subset['polarity']*100, color=color, label=category, alpha=0.6, edgecolor='k')

# Add legend manually
from matplotlib.lines import Line2D
legend_elements = [Line2D([0], [0], marker='o', color='w', markerfacecolor=color, markersize=10, label=category) for category, color in category_colors.items()]
ax.legend(handles=legend_elements, loc='upper right')

plt.title('Geospatial Visualization of Tweet Polarity')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.show()
```

The geospatial visualization of tweet polarity related to ChatGPT reveals significant insights into global sentiment. The map shows a predominance of positive sentiment, represented by green points, indicating widespread favorability towards ChatGPT. Neutral tweets, shown in yellow, are also present, suggesting balanced and factual discussions without strong emotions. Negative sentiments, marked in red, are sparse, indicating limited criticism.

Regionally, the USA exhibits a high density of tweets, especially around major cities like New York, San Francisco, and Chicago, with a strong positive reception. Europe, including the UK, Germany, and France, shows a mix of positive and neutral sentiments, reflecting engaged and favorable discussions. In Asia, countries like India, Japan, and Australia also display predominantly positive tweets, indicating favorable opinions about ChatGPT.

Other regions, including South America and Africa, have fewer tweets but still exhibit mostly positive sentiments. This widespread positivity towards ChatGPT underscores global enthusiasm and acceptance of the AI technology, while the presence of neutral tweets indicates balanced discussions. The limited negative sentiment suggests that criticisms or concerns about ChatGPT are not prevalent, highlighting its overall positive reception worldwide.
